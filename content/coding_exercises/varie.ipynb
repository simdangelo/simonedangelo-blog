```python
import asyncio
import time

async def fetch_data(url):
    print(f"Starting fetch for {url} at {time.time()}")
    # 'await' tells the event loop: "I'm waiting for I/O, you can go run other tasks."
    # asyncio.sleep() is a non-blocking equivalent of time.sleep()
    await asyncio.sleep(2)
    print(f"Finished fetching {url} at {time.time()}")
    return {"data": f"content from {url}"}

async def main():
    print("Main function started.")
    # Create a list of tasks that can be run concurrently
    tasks = [
        fetch_data("url1"),
        fetch_data("url2"),
        fetch_data("url3")
    ]
    # 'asyncio.gather' runs all tasks concurrently on the single thread
    results = await asyncio.gather(*tasks)
    print("All tasks finished.")
    print(results)

# Run the main function
asyncio.run(main())
```
